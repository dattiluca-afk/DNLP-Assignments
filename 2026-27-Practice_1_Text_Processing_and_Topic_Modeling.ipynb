# **Text Processing**
---
The text processing phase is a preliminary stage where the text to be manipulated is processed to be ready for subsequent analysis.

Text processing usually entails several steps that could possibly include:
- **Language Identification**: identifying the language of a given text.
- **Tokenization**: splitting a given text into several sentences/words.
- **Dependency tree parsing:** analyzing the dependencies between words composing the text.
- **Stemming/Lemmatization:** obtain the root form for each word in text.
- **Stopword removal**: removing words that are commonly used that they carry very little useful information.
- **Part of Speech Tagging:** given a word, retrieve its part of speech (proper noun, common noun or verb).

```python
import spacy

LOADED_MODELS = {}

def get_nlp_model(lang):
    """Safely retrieves or creates a spaCy model with English fallback."""
    lang = str(lang).strip() if lang else 'en'

    if lang in LOADED_MODELS:
        return LOADED_MODELS[lang]

    # Attempt 1: Try full pre-trained model (e.g. 'ja_core_web_sm')
    try:
        nlp = spacy.load(f"{lang}_core_web_sm")
    except (OSError, ImportError):
        # Attempt 2: Try language-specific blank model (e.g. spacy.blank('ja'))
        try:
            nlp = spacy.blank(lang)
        except (OSError, ImportError, ValueError):
            # Attempt 3: Safe fallback to English blank model (requires zero external libraries)
            try:
                nlp = spacy.load("en_core_web_sm")
            except (OSError, ImportError):
                nlp = spacy.blank("en")

    # Cache result so we don't repeat try-except logic on every row
    LOADED_MODELS[lang] = nlp
    return nlp


def spacy_tokenizer(row):
    nlp = get_nlp_model(row['language'])
    doc = nlp(str(row['Text']))
    return [token.text for token in doc]


# Apply across DataFrame rows
df['tokens'] = df.apply(spacy_tokenizer, axis=1)
```

### Corpus

A Document is a text.

A **corpus is a collection of Document objects**. Corpora can be used as:
1) Input for training a Model. During training, the **models use this training corpus to look for common themes and topics, initializing their internal model parameters.**
Gensim focuses on **unsupervised models** so that no human intervention, such as costly annotations or tagging documents by hand, is required.
2) Documents to organize. After training, a topic model can be used to extract topics from new documents (documents not seen in the training corpus).




```python
from gensim.utils import simple_preprocess

# 1. I take each headline, I convert it into tokens [token1, token2, token3, ...]
texts = [simple_preprocess(headline) for headline in covid_df['headlines'].values]
# 2. I create a Dictionary, accessible via dct.token2id
dictionary = corpora.Dictionary(texts)
dictionary.token2id
# 3. Converts each document into a list of 2-uples (token_id, number_of_appearance).
# Essentially, it creates a BoW [(token1,number_of_app1), (token2, number_of_app2), ...] for each document.
corpus = [dictionary.doc2bow(text) for text in texts]
```

```python
# 4. Trains LSI model on corpus and dictionary, on number of topic set to num_topics

from gensim.models import LsiModel
num_topics = 5
lsi_model = LsiModel(corpus=corpus, id2word = dictionary, num_topics = num_topics)
```

```python
# 5. Inspect num_topics topics and their top contributing words
topics = lsi_model.print_topics(num_topics=num_topics, num_words=10)
for topic_id, topic_words in topics:
    print(f"Topic {topic_id + 1}:\n{topic_words}\n" + "-" * 50)
```

Those output topics illustrate why raw text needs extra preprocessing before topic modeling.

Notice how the top-scoring terms across almost all 5 topics are generic English stopwords (`"the"`, `"of"`, `"in"`, `"to"`, `"and"`, `"is"`, `"that"`) along with high-frequency domain terms (`"covid"`, `"coronavirus"`). Because LSI works by identifying mathematical axes of maximum variance via SVD, common words that appear across almost every document end up dominant, making it hard to interpret what each topic actually represents.

Notice also that LSI produces **negative weights** (like `-0.672*"in"` in Topic 2). In LSI, topics are orthogonal axes in a vector space:
- A **positive weight** means the presence of that word pulls a document in the direction of that topic.
- A **negative weight** means the word pulls the document in the opposite direction along that same axis.

## Exercise 8

The top-scored words contributing to each topic (if no stopword removal is applied) are english common words (e.g., *to, for, in, of, on*..). Moreover, missing punctuation removal could be critical for topic identification. Repeat the same procedure of Ex. 7 by adding preliminary preprocessing step to:
1. **remove stopwords**
2. **strip punctuation**
3. **lowercase all words**

```python
from gensim.utils import simple_preprocess
from gensim.parsing.preprocessing import remove_stopwords, strip_punctuation

def add_prep(x):
  no_punct = strip_punctuation(x.lower())
  no_sw = remove_stopwords(no_punct)
  return no_sw

# 1. I take each headline, I convert it into tokens [token1, token2, token3, ...]
texts = [simple_preprocess(add_prep(headline)) for headline in covid_df['headlines'].values]
# 2. I create a Dictionary, accessible via dct.token2id
dictionary = corpora.Dictionary(texts)
dictionary.token2id
# 3. Converts each document into a list of 2-uples (token_id, number_of_appearance).
# Essentially, it creates a BoW [(token1,number_of_app1), (token2, number_of_app2), ...] for each document.
corpus = [dictionary.doc2bow(text) for text in texts]

# 4. Trains LSI model on corpus and dictionary, on number of topic set to num_topics

from gensim.models import LsiModel
num_topics = 5
lsi_model = LsiModel(corpus=corpus,
                     id2word = dictionary,
                     num_topics = num_topics)

# 5. Inspect num_topics topics and their top contributing words
topics = lsi_model.print_topics(num_topics=num_topics, num_words=10)
for topic_id, topic_words in topics:
    print(f"Topic {topic_id + 1}:\n{topic_words}\n" + "-" * 50)
```

From personal notes:

"Each **row** of $U_k$ represents a word in the corpus of documents, and the first $k$ columns describe the **relevance of each word to each of the top-k concepts**".


On the other hand, given any of the columns from $U_k$, **each row represents the contribution of each word to that topic.**

So *LsiModel* from *gensim* simply computes for each $k$:  $\sum_{i=1}^{\#words}w_{i,k}\times word_i$.

This function `generate_topics` is designed to make it easy for you to explore
how a language model extracts topics from sentences.  

- You do **not** need to modify the function itself.
- In the upcoming steps, all you will do is **call this function** with a new
  prompt describing the topics you want the model to extract.
- The function will:
    1. Randomly pick a few sentences from the dataset.
    2. Send them to the model with your prompt.
    3. Print each sentence along with the predicted topics.
- Make sure your prompt clearly instructs the model on what format to return.

```python
def generate_topics(prompt, model, tokenizer, sentences, n_examples=5, device='cuda'):
    """
    Generate topics for random sentences using a language model.

    Args:
        prompt (str): Instruction for the model.
        model: The language model.
        tokenizer: The tokenizer for the model.
        sentences (list of str): Sentences to select from.
        n_examples (int): Number of sentences to process.
        device (str): Device to run the model on.

    Returns:
        list of (sentence, topics) tuples.
    """

    # Ensure sentences is a standard Python list to avoid pandas indexing issues
    if hasattr(sentences, "tolist"):
        sentences = sentences.tolist()
    elif not isinstance(sentences, list):
        sentences = list(sentences)

    results = []
    for i, sentence in enumerate(random.choices(sentences, k=n_examples)):
        full_prompt = f"{prompt}{sentence}\nTopics: "
        inputs = tokenizer(full_prompt, return_tensors="pt").to(device)

        output_ids = model.generate(**inputs, max_new_tokens=32)
        decoded = tokenizer.batch_decode(output_ids, skip_special_tokens=True)[0]

        if "Topics:" in decoded:
            topics = decoded.split("Topics:")[-1].strip()
        else:
            topics = decoded.strip()

        results.append((sentence, topics))
        print(f"Sentence {i+1}: {sentence}\nTopics: {topics}\n{'-'*50}")
    return results
```


**1<sup>st</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>without providing</u> a predefined list of topics to choose from.\
*Example of prompt*:\
Which are the most relevant topics of the following sentence?

```python
PROMPT1 = "Identify the topics"
generate_topics(PROMPT1, model, tokenizer, sentences, n_examples=10, device=device)
```

**2<sup>nd</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>providing</u> a predefined list of topics to choose from.

*Example of prompt*:\
Which are the most relevant topics of the following sentence?\
Choose among: medicine, COVID, Artificial Intelligence, treatment, English literature, vaccine, gardening

```python
PROMPT2 = """Identify the topics among these: medicine, sport, entertainment"""
generate_topics(PROMPT2, model, tokenizer, sentences, n_examples=10, device=device)
```

**3<sup>rd</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>providing</u> a predefined list of topics to choose from along with the corresponding definitions.
